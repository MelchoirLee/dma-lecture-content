## Pandas

Pandas ("Python data analysis") is a Python module that contains data structures and tools for working with data.

It has a lot of numpy-like functionality in it, like vector-based operations

However, pandas specializes in dealing with tabular data composed of mixed data types, while an numpy array can usually hold a single data type (although there is a newer type of of numpy array, a structured array, that can hold more than one.)

Also, pandas has an explicit idea of named columns and an index column that indexes the rows. Numpy just has indeces and slices (which pandas has as well.)

## Types in Pandas

- Series: a one-dimensional, labeled array
- DataFrame: two dimensional data structure (with named columns, like a spreadsheet with a header row)
  - each column of a DataFrame is itself a Series
- Index

## Pandas Series

- a one-dimensional labeled array
- the associated labels are collectively referred to as the index
- Series properties:
  - values: the actual data in the Series
  - index: the labels for the data in the Series

## DataFrames

- a rectangular table of data
- can think of it also as an ordered collection of columns that have names
  - each column is a Series
  - so like a dictionary of Series objects where the column names are the keys
- DataFrame index are row names
- DataFrame columns are column names
- DataFrame data is the data in the table
- data, index, and columns can be specified when making a DataFrame
