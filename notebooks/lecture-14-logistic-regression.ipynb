## Predicting Probabilities

Suppose we wanted to predict a probability, which must between zero and one.

We could use linear regression. The trouble with doing this, even if the model is trained on data where the target variable is between 0 and 1, it can still predict values outside that interval. One solution to this is to just consider any prediction which is greater than one to be one, and less than zero to be zero.

This is sometimes done, and when it is, it is called the linear probability model. But usually people use another model to predict probabilities, the logistic regression model.

## Transforming Probabilities to Log Odds

The idea of logistic regression is based on transforming a probability p to a new value that can range from −∞ to +∞.

We use the following formula to do that:

$$
\ln(\frac{p}{1-p})
$$

where ln signifies the natural logarithm (log base 𝑒).

𝑝1−𝑝 is the odds of an event with probability. For instance, if p is 0.75, 1-p is 0.25 and the odds are 3 (often written as 3 to 1). The odds can be very high (say 999 to 1) or very low (1 to 999) and when we apply the log function to these high or low numbers we get positive or negative numbers.

## A Table of Odds and Log Odds

The following code shows the odds and the log odds for probabilities from 10 to the minus 10 up through 0.1, for a probability of 0.5, and for probabilities of 0.9, 0.99, etc. up to 0.999999999. As we can see, the log odds occupies the entire real line in principle, although its growth rate (in magnitude) is slow.

```python
import math
print(f"{'Probability':15s}{'1-Probability':15s}{'Odds':15s}{'Log Odds':15s}")
for i in range(-10,11,1):
  if i==0:
    odds=1
    print(f"{0.5:.12f} {0.5:.12f} {odds:.12f} {math.log(odds):.3f}")
    continue
  if i<0:
    p=10**i
  else:
    p=1-10**(-i)
  odds=p/(1-p)
  if i<0:
    print(f"{p:.12f} {1-p:.12f} {odds:.12f} {math.log(odds):.3f}")
  else:
    print(f"{p:.12f} {1-p:.12f} {odds:14.2f} {math.log(odds):.3f}")
```

## The Relationship Between Linear and Logistic Regression

Logistic regression is simply linear regression after the target variable, which initially ranges from 0 to 1 (like a probability or a percentage) is transformed to log odds. So we simply have (if we have 𝑚 predictors:

$$
\ln \frac{p}{1-p} = \beta_0+\beta_1x_1+\beta_2x_2+\cdots+\beta_mx_m
$$

which transforms to (by exponentiating both sides and re-arranging).

$$
p = \frac{1}{1+e^{-(\beta_0 + \beta_1 x_1 + \beta_2 x_2 + \cdots + \beta_mx_m )}}
$$

The function on the right is called the logistic function; the formula shows the logistic function applied to of the linear combination of predictors and betas. The logistic function has a maximum value of one (in the case when the linear combination of predictors becomes very large and positive) and a minimum value of zero (when the linear combination of predictors becomes large in magnitude but negative).

Note in the case of a single predictor, 𝛽<sub>0</sub> becomes 𝛼 and 𝛽<sub>1</sub> simply 𝛽.

Unlike simple linear regression, there is no closed form for estimating the parameters. The parameters (the betas) are usually estimated using a technique called [maximum likelihood estimation](https://online.stat.psu.edu/stat415/lesson/1/1.2) (MLE).

## Concepts in Binary Classification

- True negative: a zero that was correctly predicted as a zero (here, if there was no default, and the model predicted no default)
- True positive: a one that was correctly predicted as a one (here, if there was a default, and the model predicted a default)
- False negative: a one that was incorrectly predicted as a zero (here, if there was a default, and the model predicted no default)
- False positive: a zero that was incorrectly predicted as a one (here, if there was no default, and the model predicted a default)

The "true" ones are when the model got it right, and the "false" ones when it got it wrong.

- True negatives are actually negatives.
- True positives are actually positives.
- False negatives are actually positives.
- False positives are actually negatives.

The goal of any model is to minimize the latter two things, if possible.

### Accuracy, Precision and Recall

#### Accuracy

Three of the metrics of model performance are accuracy, precision, and recall.

The first metric, accuracy, is the percent of the time that the model was correct overall.

Another way to put this is that is the sum of the true negatives and the true positives divided by the total size of the dataset.

Since it is wrong in 63 cases out of the 2500 in the test set, that comes out to 97.4% accuracy.

What would the accuracy be if the model always picked no default (zero)? Remember only three percent of the cases are defaults.

#### Precision

Precision is defined with respect to the total number of times the model predicted a positive result. It is the share of those predictions that are correct.

Another way to say this is that it is the true positives divided by the sum of the true positives and false positives. Ideally, there would be no false positives.

So here it is the share of correct default predictions divided by all default predictions. That is 22 out of 29, or 75.9 percent.

#### Recall

Recall is defined as the share of the true positive values detected by the model, that is the true positives divided by the sum of the true positives and the false negatives. This is the metric on which the model does the worst. Recall is also called sensitivity, hit rate, or true positive rate (TPR)

There are 78 actual defaults and the model detected 22 of these, missing 56. That is 28.2% recall

### The Threshold

By default, when logistic regression is used for binary classification, probabilities above 0.5 are considered ones and below 0.5, zeros. Thus 0.5 is called the threshold.

What would happen to the number of positives detected when you lower the threshold? How about false positives? How about false negatives?

### The Receiver Operating Characteristic (ROC) Curve

The Receiver Operating Characteristic (ROC) curve shows the tradeoffs when you raise or lower the threshold.

It was originally developed in the context of radar in World War II. Allied radar receivers were detecting Japanese aircraft.

If you lowered their detection threshold a lot, you would detect more Japanese aircraft, but you would have more false positives, which would mean you would try to attack more planes that weren't Japanese.

However, if you raised it, you would have too many false negatives, that is, you would miss some Japanese planes.

The ROC curve for this default detection model is shown in the notebook below.

On the left of the plot, it shows that if we raise the threshold sufficiently, our chance of false alarm can be lowered to zero, but we will only detect about 20 percent of the true positives.

As we move along the blue curve, the threshold is lowered. This increases the true positive rate (recall) but at the expense of the false positive rate. We can achieve 100 percent recall, but we will need to tolerate 30-40 percent false positives. That is, 30-40 percent of the true negatives will be incorrectly classified as positive.

The area under the curve (AUC) is a measure of the performance of the classifier. Here it is 0.96.

```python
import pandas as pd 
data = pd.read_csv("default.csv")

import seaborn as sns
import warnings
warnings.filterwarnings("ignore", category=FutureWarning)

#define the predictor variable and the response variable
x = data['balance']
y = data['default']
```

```python
from sklearn import metrics
import statsmodels.api as sm
X = sm.add_constant(data['balance'], prepend=False)
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=123)
model = LogisticRegression()
model.fit(X_train, y_train)
```

```python
y_pred = pd.Series(model.predict(X_test))
y_test = y_test.reset_index(drop=True)
z = pd.concat([y_test, y_pred], axis=1)
z.columns = ['True', 'Prediction']
z[(z['Prediction']==1) & (z['True']==0)]
```

```python
print("Accuracy:", metrics.accuracy_score(y_test, y_pred))
print("Precision:", metrics.precision_score(y_test, y_pred))
print("Recall:", metrics.recall_score(y_test, y_pred))
```

```python
z[z['Prediction']==1]['True'].value_counts()
```

```python
import sklearn.metrics as metrics
# calculate the fpr and tpr for all thresholds of the classification
probs = model.predict_proba(X_test)
preds = probs[:,1]
fpr, tpr, threshold = metrics.roc_curve(y_test, preds)
roc_auc = metrics.auc(fpr, tpr)

import matplotlib.pyplot as plt
plt.title('Receiver Operating Characteristic')
plt.plot(fpr, tpr, 'b', label = 'AUC = %0.2f' % roc_auc)
plt.legend(loc = 'lower right')
#plt.plot([0, 1], [0, 1],'r--')
plt.xlim([0, 1])
plt.ylim([0, 1])
plt.ylabel('True Positive Rate (Recall)')
plt.xlabel('False Positive Rate (Chance of False Alarm)')
plt.show()
```
