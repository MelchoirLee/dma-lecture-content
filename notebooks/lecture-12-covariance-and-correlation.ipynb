## Covariance and Correlation

There are two measures of how two variables vary together: covariance and correlation. Covariance is unnormalized and correlation is normalized. Unnormalized means that the covariance depends on the scale of the two variables, and normalized means that the correlation doesn't.

In the following formula for the covariance, E is the expected value, which is the population mean. Intuitively, if each variable is on the same side of its mean at the same time as the other, they have a positive covariance, if they are are different sides, they have a negative covariance, and if they are unrelated, they have a covariance of zero or close to zero.

$$
cov_{X,Y}=\frac{\sum_{i=1}^{N}(X_{i}-E(X))(Y_{i}-E(Y))}{N}
$$

## Correlation

The magnitude of the covariance is hard to interpret and depends on the magnitudes of the variables in question. So in practice, people tend to use a normalized version of the covariance, which is the correlation coefficient. It has the same sign as the covariance, but ranges between -1 and +1. It is the covariance divided by the product of the standard deviations of the two variables.

$$
\rho = \frac{\text{cov}(X,Y)}{\sigma_x \sigma_y}
$$

This code shows a plot of the correlation matrix. Note that the diagonal and the upper part of the matrix are suppressed. This is because the numbers above the diagonal are the same as below (the correlation of x and y is the same as that of y and x) and because the numbers on the diagonal are one (because the correlation of a variable with itself is one.)

We can see from this that there is a strong correlation between height and weight and a weaker (and negative) one between gender and weight and gender and height. This is because the gender variable is coded one for female and women tend to have a lower height and weight on average than men.

```python
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
weight_height= pd.read_csv('weight-height.csv')
weight_height['Female']=weight_height['Gender'].apply(lambda x: 1 if x=="Female" else 0)
weight_height.drop('Gender',axis=1, inplace=True)
corr=weight_height.corr()

sns.set_theme(style="white")

# Generate a mask for the upper triangle
mask = np.triu(np.ones_like(corr, dtype=bool))

# Set up the matplotlib figure
f, ax = plt.subplots(figsize=(11, 9))

# Generate a custom diverging colormap
cmap = sns.diverging_palette(230, 20, as_cmap=True)

# Draw the heatmap with the mask and correct aspect ratio
sns.heatmap(corr, mask=mask, cmap=cmap, vmax=1, center=0,
            square=True, linewidths=.5, cbar_kws={"shrink": .5})
plt.show()
```

Here we generate some normally distributed data for 26 columns that we call A through Z, one column per letter of the alphabet.

We then generate a heat map of the correlation matrix. On average the correlation should be zero, but there are some correlations that are non-zero because of the randomness.

```python
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
weight_height= pd.read_csv('weight-height.csv')
weight_height['Female']=weight_height['Gender'].apply(lambda x: 1 if x=="Female" else 0)
corr=weight_height.corr(numeric_only=True)

from string import ascii_letters
sns.set_theme(style="white")

# Generate a large random dataset
rs = np.random.RandomState(33)
d = pd.DataFrame(data=rs.normal(size=(100, 26)),
                 columns=list(ascii_letters[26:]))

# Compute the correlation matrix
corr = d.corr()

# Generate a mask for the upper triangle
mask = np.triu(np.ones_like(corr, dtype=bool))

# Set up the matplotlib figure
f, ax = plt.subplots(figsize=(11, 9))

# Generate a custom diverging colormap
cmap = sns.diverging_palette(230, 20, as_cmap=True)

# Draw the heatmap with the mask and correct aspect ratio
sns.heatmap(corr, mask=mask, cmap=cmap, vmax=.3, center=0,
            square=True, linewidths=.5, cbar_kws={"shrink": .5})
plt.show()
```
