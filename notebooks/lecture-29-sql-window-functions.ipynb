## Window Functions and Aggregation Functions Do Different Things

<div style="text-align:center;margin:0.4em 0;"><img src="images/DZBrPEA0K4qUQXS1xhr8oU8i.png" alt="" style="max-width:100%;width:512px;height:auto;"></div>

## Aggregation Functions

- Reduce many rows to one using an aggregate function such as sum(), count(), avg()
- Therefore typically reduce the number of rows.

## Window Functions

- Keep the same number of rows
- Calculate new columns based on groups of rows called “windows”
- Can do aggregations within windows or pull out particular values
- “Over” clause specifies window for a new variable
